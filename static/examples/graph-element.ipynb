```javascript
graph.clear()

// A class of static methods for constructing a 
Pyramid = class Pyramid {
  // constructs a line of length l, using the implied graph object.
  static line(l){
    const vs = [];
    let id, prevId;
    
    for(let i=0; i<=l; i++){
      vs.push(id = graph.addVertex().id);
      if(prevId) graph.addEdge(prevId, id);
      prevId = id;
      
      vs.push(id);
    }

    // the return value is a line, a list of connected vertices
    return vs;
  }

  // connecting every one vertex in line A with two neighbor vertices in line B
  static stitchLinesOdd(A, B){
    for(let i=0; i<A.length; i++){
      graph.addEdge(A[i], B[i]);
      graph.addEdge(A[i], B[i+1]);
    }
  }

  // connecting one vertex in A with one vertex in B
  static stitchLinesEven(A, B){
    for(let i=0; i<A.length; i++){
      graph.addEdge(A[i], B[i]);
    }
  }

  // constructs a triangle of width w, w lines long/wide, using line() and 
  // stitchLinesEven(). 
  static triangle(w){
    const lines = [];
    let line, prevLine;
    
    for(let i=0; i<=w; i++){
      line = Pyramid.line(i);
      if(prevLine) Pyramid.stitchLinesOdd(prevLine, line);
      prevLine = line;
      
      lines.push(line);
    }

    return lines;
  }

  // connecting every ith line in triangle A evenly the ith line in triangle B,
  // as well as oddly the i+1th line in triangle B.
  // A and B are triangles, ie. lists of lines.
  static stitchTriangles(A, B){
    // A and B are lists of lines.
    for(let i=0; i<A.length; i++){
      Pyramid.stitchLinesEven(A[i], B[i]);
      Pyramid.stitchLinesOdd(A[i], B[i+1]);
    }
  }

  // constructs a pyramid of height h, using triangle() and stitchTriangle().
  static pyramid(h){
    const triangles = [];
    let triangle, prevTriangle;

    for(let i=0; i<h; i++){
      triangle = Pyramid.triangle(i);
      if(prevTriangle) Pyramid.stitchTriangles(prevTriangle, triangle);
      prevTriangle = triangle;
      
      triangles.push(triangle);
    }

    return triangles;
  }
}

Pyramid.pyramid(5)
```
