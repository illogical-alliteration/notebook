# Graph Element

It has been a long and fulfilling obsession of mine to understand and implement my own version of an academic paper that hit the arxiv preprint server in 2007: Dynamic Multilevel Graph Visualization (Veldhuizen, 2007). 

I've worked on many versions over the years, always learning something new that carried over to the next iteration, sometimes going in directions I saw no other way out than to start over with a fresh slate. Alas, I have yet to complete the whole thing. All of the libraries were built on top of mrdoob's ThreeJS, and I have built small applications making use of them. 

But I particularly proud of the latest version, published under @verticesandedges/graph-element on npm, which means we can pull it into this notebook and play with graphs. 

When you execute the javascript cell below, it creates a graph canvas, which you wil control in the next code cell after that. 

```javascript
const { GraphElement } = await import("https://esm.sh/@verticesandedges/graph-element/dist/index.js");

graph = document.createElement('graph-el');
graph.style.height = "500px";
graph.style.border = '1px solid lightgray';
return graph
```

## Adding Elements
Now that we have a graph, there are four main things we can do. We can add and remove vertices and edges. And set their appearances, either when you create them or before you remove them. 

```javascript
// adding elements
a = graph.addVertex();
b = graph.addVertex();

e = graph.addEdge(a, b);
```

## Removing Elements

And this is how to remove elements from the screen:

```javascript
// severs the edge between a and b
graph.removeEdge(e);
```

```javascript
// removes a
graph.removeVertex(a);
```

## Changing Elements

And finally, the things you can change about a vertex:

* size (number)
* color (css color string)
* texture (image file path)
* visible (boolean)
* selected (boolean)
* selection color (css color string)
* label (string)

and a couple options for the edge:

* color (css color string)
* strength (number)
* arrow (boolean)
* spline (boolean)
* width (number)
* visible (boolean)
* label (string)

These phrases are the element attributes as well, (spell them in camelCase!).

```javascript
a.color = "red";
e.spline = true;
```

```javascript
graph.addVertex({size: 25});
```
