```javascript
const { GraphElement } = await import("https://esm.sh/@verticesandedges/graph-element/dist/index.js");

const size = 5; // careful with this number, from 1 to about 10.
const vo = { size: 1.0 }; // vertex options
const eo = { width: 1.0 }; // edge options

graph = document.createElement('graph-el');
graph.style.height = "500px";

let deps = []; // depths
for(let i=0; i<size; i++){
  let cols = []; // columns
  for(let j=0; j<size; j++){
    let row = []; // rows
    for(let m=0; m<size; m++){
      let v = graph.addVertex(vo);

      // Get the previous along each dimension of the cube  
      if(m > 0) graph.addEdge(v, row[row.length - 1], eo);
      if(j > 0) graph.addEdge(v, cols[cols.length - 1][m], eo);
      if(i > 0) graph.addEdge(v, deps[deps.length-1][j][m], eo);
      
      row.push(v);
    }
    
    cols.push(row);
  }
  
  deps.push(cols);
}

return graph;
```
