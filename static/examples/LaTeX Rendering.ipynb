## Introduction

I recently completed LaTEX rendering within markdown cells. The magic behind the input boxes is codemirror, a latex translator, and some gluecode helpfully supplied by Google's Gemini. The custom code involved a Plugin extension for the codemirror view. In the following two cells, you'll see code and result. (If the two cells look the same, double click one of them to see the source and its result at the same time. or click Render). 

### Pairs of $ 
Use a pair of $-symbols to wrap math showing on the same line as your surrounding text.

Text before $x = \frac{-b\pm\sqrt{b^2-4ac}}{2a}$ Text after

### Pairs of $$
Similarly, 
<strong>$$</strong> math <strong>$$</strong>
renders the math as a figure, but in the middle of its own line. 


Text before
$$
x = \frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$
Text after
